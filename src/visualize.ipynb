```python
import matplotlib.pyplot as plt
import torch

from model import ImprovedCNN
```

```python
model = ImprovedCNN()

x = torch.randn(
    1,3,32,32
)

features = model.features[0](x)[0].detach()


fig, axes = plt.subplots(
    4,8,
    figsize=(10,8)
)


for i, ax in enumerate(axes.flat):

    if i < features.shape[0]:
        ax.imshow(
            features[i],
            cmap="gray"
        )

    ax.axis("off")


plt.suptitle(
    "First Convolution Feature Maps"
)

plt.savefig(
    "results/features.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
